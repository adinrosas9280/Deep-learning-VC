**Brillo y contraste**
Simple operación aritmética **punto a punto** (no espacial):

$$ I_{salida}(x,y) = \alpha \cdot I_{entrada}(x,y) + \beta $$

donde $\alpha$ controla el contraste y $\beta$ controla el brillo

```python
def ajustar_brillo_contraste(img, alpha=1.0, beta=0):
    """alpha: contraste (1.0 = sin cambio) | beta: brillo (0 = sin cambio)"""
    ajustada = img.astype(np.float32) * alpha + beta
    ajustada = np.clip(ajustada, 0, 255).astype(np.uint8)
    return ajustada

mas_brillo   = ajustar_brillo_contraste(imagen_rgb, alpha=1.0, beta=60)
menos_brillo = ajustar_brillo_contraste(imagen_rgb, alpha=1.0, beta=-60)
mas_contraste = ajustar_brillo_contraste(imagen_rgb, alpha=1.8, beta=0)

fig, ax = plt.subplots(1, 4, figsize=(18, 5))
ax[0].imshow(imagen_rgb); ax[0].set_title("Original"); ax[0].axis("off")
ax[1].imshow(mas_brillo); ax[1].set_title("Más brillo (+60)"); ax[1].axis("off")
ax[2].imshow(menos_brillo); ax[2].set_title("Menos brillo (-60)"); ax[2].axis("off")
ax[3].imshow(mas_contraste); ax[3].set_title("Más contraste (x1.8)"); ax[3].axis("off")
plt.show()
```

El conjunto de datos CIFAR-10 consta de 60000 imágenes de 32x32 colores (32, 32, 3) en 10 clases, con 6000 imágenes por clase. Hay 50000 imágenes de entrenamiento y 10000 imágenes de prueba.

El conjunto de datos se divide en cinco lotes de entrenamiento y un lote de prueba, cada uno con 10000 imágenes. El lote de prueba contiene exactamente 1000 imágenes seleccionadas al azar de cada clase. Los lotes de entrenamiento contienen las imágenes restantes en orden aleatorio, pero algunos lotes de entrenamiento pueden contener más imágenes de una clase que de otra. Entre ellos, los lotes de capacitación contienen exactamente 5000 imágenes de cada clase.

Las clases son completamente excluyentes entre sí. No hay superposición entre automóviles y camiones. “Automóvil” incluye sedanes, SUV, cosas por el estilo. “Camión” incluye solo camiones grandes. Tampoco incluye camionetas.

```python
import torch
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np

# 1. Definir transformación básica para convertir la imagen a Tensor PyTorch (C, H, W) valores [0 1]
transform = transforms.Compose([
    transforms.ToTensor()
])

print("Descargando dataset CIFAR-10...")
trainset = torchvision.datasets.CIFAR10(
    root='./data',
    train=True,
    download=True,
    transform=transform
)

classes = ('Avión', 'Automóvil', 'Pájaro', 'Gato', 'Ciervo',
           'Perro', 'Rana', 'Caballo', 'Barco', 'Camión')

```

**Ejercicio: Estructura del dataset**

```python
print("Tipo de dato       :" )
print("Forma del tensor    :")
print("Rango de valores    :")
print("Etiqueta (índice)   :")
print("Etiqueta (clase)    :")
```

**Extraer exactamente 1 imagen representativa por cada clase**

```python
images_per_class = {}
for image, label in trainset:
    if label not in images_per_class:
        images_per_class[label] = image
    if len(images_per_class) == 10:
        break
```

**Visualizar las 10 imágenes en una cuadrícula de 2x5**

```python

fig, axes = plt.subplots(2, 5, figsize=(14, 6))
fig.suptitle("CIFAR-10: Muestra de 1 Imagen por Clase", fontsize=16, fontweight='bold')

for label_idx in range(10):
    row = label_idx // 5
    col = label_idx % 5

    # En PyTorch el formato es (Canales, Altura, Ancho) -> Transponemos a (Altura, Ancho, Canales) para Matplotlib
    img_tensor = images_per_class[label_idx]
    img_np = img_tensor.permute(1, 2, 0).numpy()

    axes[row, col].imshow(img_np)
    axes[row, col].set_title(f"Clase {label_idx}: {classes[label_idx]}")
    axes[row, col].axis('off')

plt.tight_layout()
plt.show()

print(f"Dimensiones de cada imagen: {img_tensor.shape} (Canales x Alto x Ancho)")
```

**Ejercicio: Separación de canales, conversión a escala de grises e histograma de 4 imagenes**

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

# Asumiendo que 'images_per_class' (diccionario de tensores de imagen) y 'classes' (tupla de nombres de clase)
# ya están definidos a partir de celdas anteriores.

selected_classes = [0, 1, 2, 3]  # Seleccionar las primeras cuatro clases para el ejercicio

for class_idx in selected_classes:
    # Obtener el tensor de la imagen de images_per_class
    img_tensor = images_per_class[class_idx]

    # Convertir el tensor de PyTorch (C, H, W) a un array de NumPy (H, W, C) y escalar a 0-255.
    # Matplotlib espera uint8 para que imshow muestre los colores correctamente.
    # Las imágenes originales están en float [0,1], por lo que se multiplican por 255 y se convierten a uint8.
    img_np_rgb = (img_tensor.permute(1, 2, 0).numpy() * 255).astype(np.uint8)

    # --- Separación de Canales ---
    # Extraer los canales R, G, B. El slicing de NumPy lo maneja automáticamente.
    R_channel = img_np_rgb[:, :, 0]
    G_channel = img_np_rgb[:, :, 1]
    B_channel = img_np_rgb[:, :, 2]

    # --- Conversión a Escala de Grises ---
    # Convertir el array RGB de NumPy a escala de grises usando OpenCV.
    # Es importante especificar COLOR_RGB2GRAY ya que img_np_rgb está en formato RGB.
    img_gray = cv2.cvtColor(img_np_rgb, cv2.COLOR_RGB2GRAY)

    # --- Trazado para cada imagen ---
    # Crear una figura con 2 filas y 4 columnas para el análisis de cada imagen.
    fig, ax = plt.subplots(2, 4, figsize=(18, 9)) # figsize ajustado para una mejor visibilidad
    fig.suptitle(f"Análisis de Imagen CIFAR-10 - Clase: {classes[class_idx]}", fontsize=16, fontweight='bold')

    # Fila 1: Imagen RGB original y canales de color individuales
    ax[0, 0].imshow(img_np_rgb)
    ax[0, 0].set_title("Original (RGB)")
    ax[0, 0].axis("off")

    ax[0, 1].imshow(R_channel, cmap="Reds")
    ax[0, 1].set_title("Canal R")
    ax[0, 1].axis("off")

    ax[0, 2].imshow(G_channel, cmap="Greens")
    ax[0, 2].set_title("Canal G")
    ax[0, 2].axis("off")

    ax[0, 3].imshow(B_channel, cmap="Blues")
    ax[0, 3].set_title("Canal B")
    ax[0, 3].axis("off")

    # Fila 2: Imagen en escala de grises y su histograma
    ax[1, 0].imshow(img_gray, cmap="gray")
    ax[1, 0].set_title("Escala de Grises")
    ax[1, 0].axis("off")

    # Trazar el histograma para la imagen en escala de grises.
    # El método .ravel() aplana el array 2D en un array 1D adecuado para hist().
    ax[1, 1].hist(img_gray.ravel(), bins=256, range=(0, 255), color="gray")
    ax[1, 1].set_title("Histograma de Intensidades")
    ax[1, 1].set_xlabel("Nivel de gris")
    ax[1, 1].set_ylabel("Número de píxeles")

    # Ocultar los subplots restantes no utilizados en la segunda fila para una visualización más limpia.
    ax[1, 2].axis('off')
    ax[1, 3].axis('off')

    plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajustar el diseño para evitar que el título se superponga con suptitle
    plt.show()
```

**Ejercicio: Elige una imagen de CIFAR-10 y aplica tres combinaciones distintas de $\alpha$ y $\beta$ (ver sección 1.6). Muestra los resultados en una sola figura y comenta qué combinación conserva mejor los detalles.**

```python
import matplotlib.pyplot as plt
import numpy as np
import cv2

# Seleccionar una imagen de CIFAR-10, por ejemplo, la primera clase (Avión)
# Asegúrate de que 'images_per_class' y 'classes' estén definidos de celdas anteriores
selected_image_idx = 0
img_tensor_cifar = images_per_class[selected_image_idx]

# Convertir el tensor de PyTorch (C, H, W) a un array de NumPy (H, W, C) y escalar a 0-255.
# Esto es necesario porque la función 'ajustar_brillo_contraste' espera uint8 entre 0-255.
img_np_cifar = (img_tensor_cifar.permute(1, 2, 0).numpy() * 255).astype(np.uint8)

# Definición de la función de ajuste (copiada de la sección 1.6 si no está disponible globalmente)
def ajustar_brillo_contraste(img, alpha=1.0, beta=0):
    """alpha: contraste (1.0 = sin cambio) | beta: brillo (0 = sin cambio)"""
    ajustada = img.astype(np.float32) * alpha + beta
    ajustada = np.clip(ajustada, 0, 255).astype(np.uint8)
    return ajustada

# Aplicar tres combinaciones distintas de alfa y beta
# Combinación 1: Más brillo, poco cambio en contraste
img_adjusted_1 = ajustar_brillo_contraste(img_np_cifar, alpha=1.0, beta=70)

# Combinación 2: Menos brillo, más contraste
img_adjusted_2 = ajustar_brillo_contraste(img_np_cifar, alpha=1.5, beta=-30)

# Combinación 3: Más brillo, menos contraste (para ver el efecto opuesto)
img_adjusted_3 = ajustar_brillo_contraste(img_np_cifar, alpha=0.8, beta=50)

# Mostrar los resultados en una sola figura
fig, ax = plt.subplots(1, 4, figsize=(18, 5))
fig.suptitle(f"Ajuste de Brillo y Contraste para la imagen de {classes[selected_image_idx]}", fontsize=16, fontweight='bold')

ax[0].imshow(img_np_cifar)
ax[0].set_title("Original")
ax[0].axis("off")

ax[1].imshow(img_adjusted_1)
ax[1].set_title(r"Más Brillo ($\beta$=70)") # Usar r'' para raw string y \beta para beta
ax[1].axis("off")

ax[2].imshow(img_adjusted_2)
ax[2].set_title(r"Más Contraste ($\alpha$=1.5), Menos Brillo ($\beta$=-30)") # Usar r'' y \alpha \beta
ax[2].axis("off")

ax[3].imshow(img_adjusted_3)
ax[3].set_title(r"Menos Contraste ($\alpha$=0.8), Más Brillo ($\beta$=50)") # Usar r'' y \alpha \beta
ax[3].axis("off")

plt.tight_layout(rect=[0, 0.03, 1, 0.9]) # Ajustar el diseño para el título
plt.show()

print("Comentario: La combinación que mejor conserva los detalles dependerá de la imagen y del objetivo. Sin embargo, un contraste muy alto (como en la combinación 2) puede "
      "exagerar los bordes y perder la gradación de tonos, mientras que un contraste muy bajo (como en la combinación 3) puede hacer que la imagen se vea 'lavada'. "
      "El ajuste de brillo debe ser moderado para evitar la pérdida de información en las luces o las sombras.")
```
